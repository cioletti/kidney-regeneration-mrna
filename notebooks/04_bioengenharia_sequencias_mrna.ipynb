# Estudo In Silico - Etapa 4: Bioengenharia das Sequências da Terapia de mRNA
### Download das Sequências Oficiais (NCBI) e Desenho Molecular do Coquetel Regenerativo

Neste notebook executamos a bioengenharia direta:
1. Download das sequências reais de RNA humano de **SOX9** e **HNF4A** diretamente do NCBI GenBank.
2. Inspeção da Região Codificadora (CDS) e tradução para os aminoácidos terapêuticos.
3. Desenho computacional do oligonucleotídeo **siRNA anti-ACTA2** para bloquear fibrose.

```python
# 1. Instalar Biopython para manipulação de sequências nucleotídicas
!pip install --quiet biopython

from Bio import Entrez, SeqIO
from Bio.Seq import Seq
import pandas as pd

# Identificação necessária pelo servidor do NCBI (NIH)
Entrez.email = "pesquisa.regeneracao.renal@gmail.com"
print("Conectado ao servidor oficial do NCBI!")
```

### 2. Download do mRNA de SOX9 Humano (Motor de Reparo Tubular)

```python
# Código de acesso oficial do transcrito de SOX9 humano: NM_000346.4
handle_sox9 = Entrez.efetch(db="nucleotide", id="NM_000346.4", rettype="gb", retmode="text")
reg_sox9 = SeqIO.read(handle_sox9, "genbank")
handle_sox9.close()

print(f"[+] Gene: {reg_sox9.id}")
print(f"[+] Descrição: {reg_sox9.description}")
print(f"[+] Comprimento total da fita de RNA: {len(reg_sox9.seq)} nucleotídeos")
print(f"[+] Primeiros 60 nucleotídeos: {reg_sox9.seq[:60]}...")
```

### 3. Download do mRNA de HNF4A Humano (Fator de Maturação Funcional)

```python
# Código de acesso oficial do transcrito de HNF4A humano: NM_000457.6
handle_hnf4a = Entrez.efetch(db="nucleotide", id="NM_000457.6", rettype="gb", retmode="text")
reg_hnf4a = SeqIO.read(handle_hnf4a, "genbank")
handle_hnf4a.close()

print(f"[+] Gene: {reg_hnf4a.id}")
print(f"[+] Descrição: {reg_hnf4a.description}")
print(f"[+] Comprimento total da fita de RNA: {len(reg_hnf4a.seq)} nucleotídeos")
print(f"[+] Primeiros 60 nucleotídeos: {reg_hnf4a.seq[:60]}...")
```

### 4. Desenho do Oligonucleotídeo de Interferência (siRNA anti-ACTA2)
A ferramenta molecular que corta o RNA da fibrose.

```python
# Sequência alvo no mRNA de ACTA2 (alfa-SMA humano)
alvo_acta2 = "AGACGCCAUCCUGUCAGCAAU"
fita_guia_antisense = str(Seq(alvo_acta2).reverse_complement().transcribe())

print("=" * 65)
print("FORMULAÇÃO MOLECULAR DO COQUETEL TERAPÊUTICO:")
print("=" * 65)
print(f"1. mRNA 1 (SOX9)       : {len(reg_sox9.seq)} nt (Modificado com N1-metilpseudouridina)")
print(f"2. mRNA 2 (HNF4A)      : {len(reg_hnf4a.seq)} nt (Modificado com N1-metilpseudouridina)")
print(f"3. siRNA anti-ACTA2    : Duplex de 21 nt")
print(f"   - Fita Sense        : 5'- {alvo_acta2} -3'")
print(f"   - Fita Guia AntiSense: 5'- {fita_guia_antisense} -3'")
print("=" * 65)
print("[SUCESSO]: Sequências biológicas validadas prontas para síntese!")
```
